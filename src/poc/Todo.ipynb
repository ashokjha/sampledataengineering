```python
import pandas as pd
import plotly.graph_objects as go

# 1. देशों का डेटा (Geographic Coordinates के साथ)
# यहाँ हमने रूट्स डिफाइन किए हैं (जैसे USA से भारत, चीन, ब्राजील आदि)
routes_data = {
    "Origin_Country": [
        "United States",
        "United States",
        "United States",
        "United States",
        "United States",
    ],
    "Origin_Lat": [37.0902, 37.0902, 37.0902, 37.0902, 37.0902],
    "Origin_Lon": [-95.7129, -95.7129, -95.7129, -95.7129, -95.7129],
    "Dest_Country": ["India", "China", "Brazil", "United Kingdom", "Australia"],
    "Dest_Lat": [20.5937, 35.8617, -14.2350, 55.3781, -25.2744],
    "Dest_Lon": [78.9629, 104.1954, -51.9253, -3.4360, 133.7751],
    # आपका पिछला आर्थिक डेटा (मंजिल वाले देश का)
    "Population": [78.9629, 104.1954, -51.9253, -3.4360, 133.7751],
    "Per_Capita_Income_USD": [78.9629, 104.1954, -51.9253, -3.4360, 133.7751],
    "Economy_GDP_USD_Trillion": [3.39, 17.96, 1.92, 3.07, 1.68],
}

df = pd.DataFrame(routes_data)

# 2. खाली मैप बेस (Empty Map Figure) की शुरुआत करें
fig = go.Figure()

# 3. मैप पर कनेक्शन लाइन्स (Lines) जोड़ना
for i in range(len(df)):
    fig.add_trace(
        go.Scattergeo(
            locationmode="country names",
            lon=[df["Origin_Lon"][i], df["Dest_Lon"][i]],
            lat=[df["Origin_Lat"][i], df["Dest_Lat"][i]],
            mode="lines",
            line=dict(width=2, color="red"),
            opacity=0.6,
            hoverinfo="skip",  # लाइनों पर माउस ले जाने पर टूलटिप न दिखे
        )
    )

# 4. मंजिल वाले देशों पर इंटरैक्टिव डॉट्स (Markers) जोड़ना
fig.add_trace(
    go.Scattergeo(
        locationmode="country names",
        lon=df["Dest_Lon"],
        lat=df["Dest_Lat"],
        hovertext=df["Dest_Country"],
        mode="markers",
        marker=dict(
            size=10, color="blue", line=dict(width=1, color="rgba(102, 102, 102)")
        ),
        # टूलटिप में दिखने वाली कस्टम जानकारी (Hover Text)
        customdata=df[
            ["Population", "Per_Capita_Income_USD", "Economy_GDP_USD_Trillion"]
        ],
        hovertemplate=(
            "<b>%{hovertext}</b><br><br>"
            + "Population: %{customdata[0]:,.0f}<br>"
            + "Per Capita Income: $%{customdata[1]:,.0f}<br>"
            + "GDP: $%{customdata[2]:.2f} T<br>"
            + "<extra></extra>"  # डिफ़ॉल्ट ट्रेस टेक्स्ट को हटाने के लिए
        ),
    )
)

# 5. मैप का लेआउट सेट करना
fig.update_layout(
    title_text="<b>Global Interactive Connection Map with Economic Data</b>",
    title_x=0.5,
    showlegend=False,
    geo=dict(
        scope="world",
        projection_type="equirectangular",
        showland=True,
        landcolor="rgb(243, 243, 243)",
        countrycolor="rgb(204, 204, 204)",
    ),
    margin={"r": 0, "t": 50, "l": 0, "b": 0},
)

# मैप को ब्राउज़र या जुपिटर नोटबुक में दिखाएं
# fig.show()
```
