# Chord Diagram

Runnable example from the accompanying README.


```python
import holoviews as hv
import pandas as pd

# Activate HoloViews' interactive Bokeh output.
hv.extension("bokeh")

# Define source-target relationships and their strengths.
links = pd.DataFrame(
    {
        "source": ["Design", "Design", "Engineering", "Marketing"],
        "target": ["Engineering", "Marketing", "Support", "Sales"],
        "value": [12, 7, 10, 8],
    }
)
names = pd.Index(pd.concat([links["source"], links["target"]]).unique())
nodes = hv.Dataset(pd.DataFrame({"index": range(len(names)), "name": names}), "index")
lookup = {name: index for index, name in enumerate(names)}
links["source"], links["target"] = links["source"].map(lookup), links["target"].map(
    lookup
)

# Draw, save, and display the interactive chord diagram.
chord = hv.Chord((links, nodes)).opts(
    labels="name",
    cmap="Category10",
    edge_color="source",
    node_color="index",
    width=700,
    height=700,
    title="Collaboration Links",
)
hv.save(chord, "chord_diagram.html")
chord
```
