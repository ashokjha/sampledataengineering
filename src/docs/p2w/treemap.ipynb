# Treemap

Runnable example from the accompanying README.


```python
import pandas as pd
import plotly.express as px

# Define a two-level organization hierarchy and one value per team.
data = pd.DataFrame(
    {
        "department": [
            "Engineering",
            "Engineering",
            "Marketing",
            "Marketing",
            "Operations",
            "Operations",
        ],
        "team": ["Platform", "Product", "Content", "Growth", "Finance", "People"],
        "headcount": [28, 22, 12, 15, 10, 8],
    }
)

# Create nested rectangles where area and color encode headcount.
fig = px.treemap(
    data,
    path=["department", "team"],
    values="headcount",
    color="headcount",
    color_continuous_scale="Blues",
    title="Headcount by Department and Team",
)

# Save and display the interactive chart.
fig.write_html("treemap.html", include_plotlyjs=True)
fig.show()
```
